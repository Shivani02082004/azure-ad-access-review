```python
"""
Generates a mock ServiceNow-style incident export CSV for the report generator.
"""
import csv
import random
from datetime import datetime, timedelta

random.seed(7)

priorities = ["P1", "P2", "P3", "P4"]
priority_weights = [0.05, 0.15, 0.45, 0.35]
categories = ["Access Request", "Password Reset", "Hardware", "Network",
              "Application Issue", "Endpoint Security", "Email"]

sla_hours = {"P1": 4, "P2": 8, "P3": 24, "P4": 72}

today = datetime.now()

with open("incidents.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["ticket_id", "category", "priority", "opened_date",
                      "resolved_date", "resolution_hours"])

    for i in range(1, 61):
        priority = random.choices(priorities, weights=priority_weights)[0]
        category = random.choice(categories)
        opened = today - timedelta(days=random.randint(1, 14), hours=random.randint(0, 23))

        # Most tickets resolved within SLA, some deliberately breach it
        target = sla_hours[priority]
        if random.random() < 0.8:
            resolution_hours = round(random.uniform(0.5, target * 0.9), 1)
        else:
            resolution_hours = round(random.uniform(target * 1.1, target * 2), 1)

        resolved = opened + timedelta(hours=resolution_hours)

        writer.writerow([
            f"INC{100000+i}",
            category,
            priority,
            opened.strftime("%Y-%m-%d %H:%M"),
            resolved.strftime("%Y-%m-%d %H:%M"),
            resolution_hours
        ])

print("incidents.csv generated with 60 mock tickets.")
```
